# **Project Summary -**



The **“Transforming EDA to Dashboard – Glassdoor Jobs Salary Analysis”** project focuses on analyzing job market and salary data to identify trends, patterns, and insights related to job roles, salary distributions, company ratings, industries, and geographical locations. The primary objective is to transform raw job listing data into meaningful business insights through Exploratory Data Analysis (EDA) and interactive dashboard development.

The project begins with data collection and preprocessing using Python libraries such as Pandas and NumPy. The dataset is examined for missing values, duplicate records, inconsistent formats, and data quality issues. Data cleaning and transformation techniques are applied to prepare the dataset for reliable analysis.

Exploratory Data Analysis is performed using Matplotlib and Seaborn to understand salary distributions, compare compensation across job titles and locations, identify high-paying roles, and examine relationships between company ratings and salaries. Statistical summaries and visualizations help uncover significant patterns and support data-driven conclusions.

The insights obtained from EDA are then transformed into an interactive dashboard using Microsoft Power BI. The dashboard presents key performance indicators (KPIs), salary comparisons, job market distributions, company insights, and location-based analysis through charts, tables, and interactive filters. These features allow users to explore the data efficiently and compare different job market segments.

The project aims to bridge the gap between raw data analysis and business intelligence by presenting complex information in a clear, interactive, and user-friendly format. It can help job seekers understand salary trends and explore career opportunities while enabling recruiters and HR professionals to gain insights into compensation patterns and the job market.

Overall, this project demonstrates practical skills in data cleaning, exploratory data analysis, data visualization, and dashboard development. It highlights the importance of transforming raw data into actionable insights that support informed decision-making. All findings and conclusions are based on the available dataset and the variables it contains.

# **Problem Statement**


The job market contains a large volume of job listings with varying salary ranges, job titles, company ratings, industries, and geographical locations. Analyzing this information manually can be time-consuming and makes it difficult to identify salary trends, high-paying job roles, and differences in compensation across companies and locations.

The objective of the “Transforming EDA to Dashboard – Glassdoor Jobs Salary Analysis” project is to analyze Glassdoor job and salary data using Python-based Exploratory Data Analysis (EDA) techniques and transform the findings into an interactive dashboard using Microsoft Power BI. The project focuses on identifying salary patterns, comparing job roles, examining company ratings, exploring geographical salary variations, and understanding the distribution of job opportunities.

The project aims to convert raw job market data into meaningful visualizations and actionable insights that help job seekers make informed career decisions and assist recruiters and HR professionals in understanding compensation trends. By integrating data cleaning, analysis, visualization, and dashboard development, the project provides an efficient and user-friendly approach to exploring job market information and supporting data-driven decision-making.

# ***Power bi dashboard overview***


1. Total Jobs and Salary Cards (KPI Cards)
Total Jobs
COUNT
Count of job listings
Average Salary
AVG
Mean salary value

Purpose: KPI cards provide a quick summary of the job market before users explore the detailed charts.
- Total Jobs: Displays the total number of job listings available in the dataset.
- Average Salary: Displays the average salary across the available job listings.

2.Clustered Bar chart
Chart type: Clustered bar chart
Purpose: To compare average salaries across different job titles and identify the roles offering the highest compensation.
Fields to use:
- Y-axis: Job Title
- X-axis: Average Salary
- Tooltips: Number of jobs and company name, if available
Expected insights:
- Which job titles have the highest average salaries?
- Which roles have comparatively lower salaries?
- How does compensation vary across job categories?
Business impact: Helps job seekers compare career options and understand salary differences between roles. For readability, display the top 10–15 job titles by average salary.

3. Pie chart
Purpose: To show how job listings are distributed across categories, such as job titles, employment types, or industries, depending on your dataset.
Fields to use:
- Legend: Job Category or Employment Type
- Values: Count of Job Listings
Expected insights:
- Which category contributes the largest share of job listings?
- Which categories have fewer listings?
- How is the job market distributed among the available categories?
Business impact: Helps users understand which job categories dominate the dataset and where job opportunities are concentrated.
Tip: If your dataset contains many unique job titles, use a broader job category or employment type instead; a pie chart with too many slices becomes difficult to read.

4. Sector wise count chart
Chart type: Clustered bar chart or column chart
Purpose: To compare the number of job listings across different sectors or industries.
Fields to use:
- Axis: Sector or Industry
- Values: Count of Job Listings
DAX measure:
Sector Job Count = COUNTROWS('Glassdoor')
Use Sector or Industry on the chart axis and the Sector Job Count measure as the value. Exclude blank sectors if appropriate.
Expected insights:
- Which sectors have the highest number of job listings?
- Which sectors have fewer opportunities in the dataset?
- How are job opportunities distributed across industries?
Business impact: Helps job seekers identify sectors with more listed opportunities and supports recruiters in understanding the distribution of demand across industries.

##  Exploratory Data Analysis (EDA) Insights

Exploratory Data Analysis (EDA) was performed on the Glassdoor Jobs Salary Analysis dataset to understand its structure, identify data quality issues, explore salary distributions, and uncover patterns in job listings across different job titles and industry sectors. The analysis provides a foundation for developing an interactive Power BI dashboard and deriving meaningful insights from job market data.

### 1. Job Listings Analysis
The total number of job listings was examined to understand the size and coverage of the dataset. Analyzing job listings by category and sector helps identify areas with a higher concentration of opportunities and sectors with fewer listings.

### 2. Salary Distribution Analysis
Salary data was analyzed using descriptive statistics and visualizations to understand average salary levels, salary variation, and the overall distribution of compensation. This analysis can help identify differences in salary levels and potential extreme values that require further investigation.

### 3. Job Title vs. Salary Analysis
Average salaries were compared across job titles to examine how compensation varies between roles. This comparison can help identify higher-paying and lower-paying job titles represented in the dataset and provide insights into salary differences across career options.

### 4. Sector-wise Job Distribution
Job listings were grouped by sector to understand how opportunities are distributed across industries. Comparing sector-wise job counts helps identify sectors that contribute the largest number of listings within the dataset.

### 5. Job Category Distribution
The distribution of listings across available job categories was examined to understand which categories account for a larger or smaller share of the dataset. These findings can be represented through a pie chart or other suitable visualization.

### 6. Data Quality and Preprocessing
The dataset was inspected for missing values, duplicate records, inconsistent formats, and incorrect data types. Data cleaning and preprocessing are important to improve the reliability of salary calculations, job counts, and dashboard visualizations.

### 7. Dashboard-Oriented Insights
The findings from EDA were used to identify the most useful metrics and visualizations for the Power BI dashboard. Total job listings and average salary provide an overview of the dataset, while the salary comparison bar chart, job distribution pie chart, and sector-wise job count chart allow users to explore different aspects of the job market.

### Conclusion
The EDA process provides a structured understanding of the Glassdoor job listings and salary dataset. It supports the identification of salary patterns, job distribution trends, and data quality issues. These findings form the analytical foundation for the Power BI dashboard, transforming raw job data into clear visual information that can support data-driven career and recruitment decisions.
